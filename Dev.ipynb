```python
import numpy as np
from prbs import PRBS
```

##### Parametros: 
- prbs_taps: Taps for the prbs poly as a list. For example: PRBS3=[3,2]
- paralellism: Number of bits per sample.
- seed_taps: Taps that define the seed for the PRBS. the length should be equal to the degree of the PRBS meaning that the MSB should be in index 2, for example, for PRBS3: seed_taps = [2,0]

```python
prbs_taps = [3,2]
seed_taps = [2,0] # or [2,1] , [2,1,0]. No other options
parallelism = 3
prbs = PRBS.create(prbs_taps,parallelism,seed_taps)
```

```python
print(prbs.get_sample())
```
